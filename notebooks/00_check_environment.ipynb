# 00 — Check the environment

Run this first. It verifies the package imports, the data bundle is reachable and
every part the figures need is present. If this notebook runs clean, the others will.


```python
import os, sys, pathlib

# repo root, whether the notebook is run from notebooks/ or from the root
ROOT = pathlib.Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

# released data: the summary bundle (all notebooks) and the raw per-pass predictions
# (notebooks 04 and 06). Set the variables before starting Jupyter to use other locations.
os.environ.setdefault('UQFIG_BUNDLE', str(ROOT / 'figure_data_bundle'))
os.environ.setdefault('UQFIG_PREDS', str(ROOT / 'cluster_preds'))
OUT = ROOT / 'out'; OUT.mkdir(exist_ok=True)

from uq_genomics.figures import style as S
S.apply_rc()
print('bundle :', os.environ['UQFIG_BUNDLE'])
print('preds  :', os.environ['UQFIG_PREDS'])
print('output :', OUT)
```

## Bundle contents

Each part is one pickled dictionary of model outputs. `comparison_*` holds the
predictive means and standard deviations per (seed, setting, method).


```python
from uq_genomics.figures import config_paths as CP, data_access as DA

parts = ['comparison_rbp','comparison_tf','comparison_gex',
         'labels_rbp','labels_tf','labels_gex',
         'classdist','sensitivity','virus','app27','umap']
for p in parts:
    b = DA.bundle_part(p)
    print(f'{p:18s}', 'OK' if b is not None else 'missing')
```
