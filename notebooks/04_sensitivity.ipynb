# 04 — Sensitivity to sources of uncertainty

Does a method's uncertainty respond when a single source of uncertainty is
introduced into the training data, and does it respond for the right classes?

Two perturbations on the simulated TF data (label noise, Fig. 9; missing motifs,
Fig. S8) and class imbalance on both real use cases (Fig. 10). Every perturbed
setting is evaluated on the test set of its reference setting, and the test sets of
the five seeds are pooled before the densities are drawn.

Two score definitions are used, and the file names say which:

* `*_pooled` — SD of each class's predicted probability over all test samples (TF).
* `*_predclass_pooled` — SD at the predicted class: for class c, the samples the
  model predicts as c (class imbalance).

The RBP panel of Fig. 10 needs the raw per-pass predictions (`UQFIG_PREDS`,
`cluster_preds/rbp_v19/`), because both RBP settings are evaluated on the same
balanced test set and that evaluation is not part of the summary bundle. Without
them the panel is skipped and the single-cell panel still builds.

## Does the depleted class move most? (Fig. S11)

The claim is not merely that the distribution moves, but that it moves most for the
classes the imbalance removed. A negative Spearman correlation between a class's
change in training-set size and its uncertainty shift means the classes that lost
training data became the more uncertain ones.
